## Data Preprocessing

```python
# Separate text column first
text_train = X['comment']
text_test = test_df['comment']

X = X.drop(columns=['comment'])
test_df = test_df.drop(columns=['comment'])

# Handle Missing Values
# Numerical → median
for col in num_cols:
    median = X[col].median()
    X[col] = X[col].fillna(median)
    test_df[col] = test_df[col].fillna(median)

# Categorical → "missing"
for col in cat_cols:
    if col in X.columns:
        X[col] = X[col].fillna("missing")
    if col in test_df.columns:
        test_df[col] = test_df[col].fillna("missing")

# Reduce Rare Categories
# If a category appears only 1% of the time, it: adds noise,increases dimensionality during encoding and may cause overfitting
for col in cat_cols:
    freq = X[col].value_counts(normalize=True)            # normalize=True : Instead of counts, it returns proportions (percentages)
    rare_categories = freq[freq < 0.01].index

    X[col] = X[col].replace(rare_categories, "rare")
    test_df[col] = test_df[col].replace(rare_categories, "rare")


# OneHot Encode for categorical columns
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)         # handle_unknown='ignore' : If the test data contains a category not seen in training, the encoder will not throw an error. Instead, it fills that row with zeros.
ohe.fit(X[cat_cols])

X_ohe = pd.DataFrame(ohe.transform(X[cat_cols]),columns=ohe.get_feature_names_out(cat_cols),index=X.index)
test_ohe = pd.DataFrame(ohe.transform(test_df[cat_cols]),columns=ohe.get_feature_names_out(cat_cols),index=test_df.index)

# Drop original
X = X.drop(columns=cat_cols)
test_df = test_df.drop(columns=cat_cols)

# Add encoded columns
X = pd.concat([X, X_ohe], axis=1)
test_df = pd.concat([test_df, test_ohe], axis=1)


# Encode target
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)


# Scale numerical columns
scaler = StandardScaler()
X[num_cols] = scaler.fit_transform(X[num_cols])
test_df[num_cols] = scaler.transform(test_df[num_cols])

text_train = text_train.fillna("")
text_test = text_test.fillna("")

# TF-IDF for comment text
# converts text into numeric vectors
tfidf = TfidfVectorizer(max_features=150000,       # max_features=150000 : Limits the vocabulary to the 150,000 most important words
                        ngram_range=(1,3),
                        min_df=2,max_df=0.95,
                        sublinear_tf=True,         # Uses log scaling for term frequency.
                        stop_words='english')      # Removes common English stopwords like:the,is,and,of,to

X_text = tfidf.fit_transform(text_train)           # X_text → sparse matrix
test_text = tfidf.transform(text_test)

# Combine text + numeric/categorical features
X_final = hstack([X_text, csr_matrix(X)])                      # hstack : It combines matrices side-by-side (columns)
test_final = hstack([test_text, csr_matrix(test_df)])          # csr_matrix(X) : Converts the DataFrame X into a sparse matrix format.

print("Final train shape:", X_final.shape)
print("Final test shape:", test_final.shape)
```

```python
# Train Validation Split
X_train, X_val, y_train, y_val = train_test_split(X_final, y_encoded, test_size=0.2, random_state=42, stratify=y)
print("Training shape: ",X_train.shape)
print("Validation shape: ",X_val.shape)
```

## Model 1: Logistic Regression

```python
results = {}              # Dictionary to store results
```

```python
lr_param_grid = {'C': [0.5,1,2,4,8]}           # C controls regularization strength in logistic regression.

lr = LogisticRegression(max_iter=2000,solver='lbfgs',n_jobs=-1,class_weight='balanced')       #class_weight='balanced' : Instead of treating all samples equally, it gives more weight to minority classes.

lr_search = GridSearchCV(lr,lr_param_grid,
                               scoring='f1_macro',
                               cv=3,n_jobs=-1
                              )

lr_search.fit(X_train, y_train)

best_lr = lr_search.best_estimator_
y_val_pred_lr = best_lr.predict(X_val)

lr_score = f1_score(y_val,y_val_pred_lr,average='macro')

# Store results
results['Logistic Regression']={'val_score':lr_score,'model':best_lr}

print("LOGISTIC REGRESSION\n")
print(f"Validation F1 Score: {lr_score}")
```

## Model 2: Ridge Classifier

```python
rc_param_grid = {'alpha': [0.1, 0.5, 1, 2, 5]}

rc = RidgeClassifier()

rc_search = GridSearchCV(rc,rc_param_grid,
                    scoring='f1_macro',
                    cv=3,n_jobs=-1
                   )

rc_search.fit(X_train, y_train)

best_rc = rc_search.best_estimator_
y_val_pred_rc = best_rc.predict(X_val)

rc_score = f1_score(y_val,y_val_pred_rc,average='macro')

# Store results
results['Ridge Classifier']={'val_score':rc_score,'model':best_rc}

print("Ridge Classifier\n")
print(f"Validation F1 Score: {rc_score}")
```

## Model 3: SGD Classifier

```python
svm_param_grid = {'loss': ['hinge','log_loss'],
    'alpha': [1e-6,1e-5,1e-4],
    'penalty': ['l2', 'elasticnet']}

svm = SGDClassifier(max_iter=2000,random_state=42)

svm_search = RandomizedSearchCV(svm,svm_param_grid,
                                scoring='f1_macro',
                                n_iter=10,
                                cv=3,n_jobs=-1,
                               )

svm_search.fit(X_train, y_train)

best_svm = svm_search.best_estimator_
y_val_pred_svm = best_svm.predict(X_val)

svm_score = f1_score(y_val,y_val_pred_svm,average='macro')

# Store results
results['SGD Classifier']={'val_score':svm_score,'model':best_svm}

print("SGD Classifier\n")
print(f"Validation F1 Score: {svm_score}")
```

## Model Comparision

```python
best_model_name = None
best_score = -1

for model in results:
    score = results[model]['val_score']
    if score > best_score:
        best_score = score
        best_model_name = model

print("Best Model:", best_model_name)
print("Validation Score:", best_score)
```
